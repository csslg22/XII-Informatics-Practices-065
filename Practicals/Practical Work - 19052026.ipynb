```python
import pandas as pd
sales_data = {
    "Category": ["Electronics", "Furniture", "Electronics", "Stationery", "Furniture"],
    "Item_Name": ["Laptop", "Chair", "Mouse", "Notebooks", "Desk"],
    "Expenditure": [45000, 12000, 1500, 2500, 18000],
}
quarterly_sales = pd.DataFrame(sales_data)
print("Original Quarterly Sales DataFrame:")
print(quarterly_sales)
print()
grouped_expenditure = quarterly_sales.groupby("Category")["Expenditure"].sum()
print("Total Expenditure Per Category:")
print(grouped_expenditure)
print()
print("Count of each Category:")
print(quarterly_sales.groupby("Category")["Category"].count())
print()
print("Find the maximum expenditure incurred by each category:")
print(quarterly_sales.groupby("Category")["Expenditure"].max())
```

```python
import pandas as pd
emp_data = {
    "EmpID": [101, 102, 103, 102, 104, 101],
    "Name": ["Anil", "Sunita", "Rajesh", "Sunita", "Megha", "Anil"],
    "Department": ["HR", "IT", "Sales", "IT", "HR", "HR"],
}
df_emp = pd.DataFrame(emp_data)
print("Original DataFrame with Duplicates:")
print(df_emp)
print()
duplicate_rows = df_emp[df_emp.duplicated()]
print("Duplicate rows identified in the dataset:")
print(duplicate_rows)
print()
cleaned_df = df_emp.drop_duplicates()
print("Cleaned DataFrame after removing duplicate rows:")
print(cleaned_df)
```

```python
import pandas as pd
data = {
    "CName": ["Jeans", "Jeans", "Shirt", "Trouser", "T-Shirt"],
    "Size": ["L", "XL", "XL", "L", "XL"],
    "Price": [1200, 1350, 900, 1000, 600]
}
row_indices = ["C1", "C2", "C3", "C4", "C5"]
cloth = pd.DataFrame(data, index=row_indices)
print("--- Original DataFrame ---")
print(cloth)
print()
cloth["CName"] = cloth["CName"].replace({"Trouser": "Pant", "Jeans": "Denim"})
print("--- a) After replacing item names ---")
print(cloth)
print()
cloth["Price"] = cloth["Price"] * 1.10
print("--- b) After increasing prices by 10% ---")
print(cloth)
print()
cloth.index = ["C001", "C002", "C003", "C004", "C005"]
print("--- c) After renaming row indexes ---")
print(cloth)
print()
# cloth = cloth.drop('C003')
cloth.drop("C003",inplace=True)
print("--- d) After deleting row C003 ---")
print(cloth)
print()
cloth.drop("Size", axis=1,inplace=True)
print("--- e) Final DataFrame after deleting 'Size' column ---")
print(cloth)
```
