```python
a = [12,14,19,24,21]
b = [10,20,30,40,50]

x = [33,44,66,55,22]
y = [15,25,35,45,55]

plt.plot(a,b,marker="o",label="Plot 1")
plt.plot(x,y,marker="s",label="Plot 2")

plt.xlabel("Values of x-axis")
plt.ylabel("Values of y-axis")
plt.legend(loc="upper left")
plt.title("Sample Line Chart")

plt.plot()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.arange(0.,10.,0.1)
a = np.cos(x)
b = np.sin(x)

plt.plot(x,a,'b')
plt.plot(x,b,'r')
plt.show()
```

```python
import pandas as pd

data = {
    'Name'     : ['Amit','Bhavna','Chintu','Deepa','Esha',
                   'Faisal','Gaurav','Hina','Ishan','Sara','Ravi'],
    'Class'    : [11, 12, 11, 11, 12, 11, 12, 11, 12, 12, 11],
    'Section'  : ['A','B','A','C','A','B','A','B','C','A','C'],
    'Marks'    : [45, 85, 30, 92, 55, 62, 78, 25, 48, 95, 60],
    'Physics'  : [50, 80, 20, 95, 60, 65, 80, 30, 40, 90, 55],
    'Chemistry': [40, 85, 35, 90, 50, 70, 75, 20, 55, 92, 65],
    'English'  : [60, 75, 40, 88, 65, 55, 85, 45, 60, 89, 70],
    'Remarks'  : ['Average','Good','Poor','Excellent','Average',
                   'Average','Good','Poor','Average','Excellent','Average']
}
df = pd.DataFrame(data)
```

```python
df
```

```python
plt.plot(df['Name'],df['Physics'],marker="s",label="Name-Phy")
plt.plot(df['Name'],df['Chemistry'],marker="d",label="Name-Chem")
plt.plot(df['Name'],df['English'],marker="o",label="Name-Eng")
plt.legend()
plt.grid()
plt.show()
```
