# Addition of Data into a DataFrame

Consider a DataFrame `Marks`

```python
import pandas as pd
data = {'Name':['Ram', 'Sonu', 'Rita'],
        'Eng':[25,23,24],
        'Maths':[29,27,26]}
Marks = pd.DataFrame(data)
print(Marks)
```

```python
# Add a new column
Marks['IP'] = [29,28,27]
print(Marks)
```

```python
Marks['Eng'] = 30
print(Marks)
```

Here, `Eng` is a column which already exists in the DataFrame `Marks`.

If we put a scalar value for an already existing column, Python will update the entire column with the new value, as seen in the example.

If, we give a column name which does not exist in the DataFrame, then Python will create a new column with similar value set for all the rows.

Example: `Marks['CS'] = 29` will add a new column CS with value 29 set for all the rows.

```python
print(len(Marks))
```

```python
Marks.loc[len(Marks)] = ['Dinesh', 29, 28, 27]
print(Marks)
# OR
# Marks[len(Marks)] = ['Dinesh', 29, 28, 27]
#print(Marks)
```

If we give an index value which does not exist, Python will insert the new row in that index position. If we give an index which already exists, Python will update the existing row.

Example:

```python
Marks.loc[3] = ['Shyam', 21, 22, 23]
print(Marks)
```

# Renaming rows & columns of a DataFrame

```python
# Let us consider the following DataFrames df1 & df2:

d1 = {'RollNo':[101,102,103],
      'Name':['Abhi','Shyam', 'Ronit']}
d2 = [[121, 'Anand'],
      [122, 'Mohit']]
df1 = pd.DataFrame(d1,index=['A1','A2','A3'])
df2 = pd.DataFrame(d2)
print("DataFrame df1:")
print(df1)
print("-"*20)
print("DataFrame df2:")
print(df2)
```

```python
# Rename single column of a labelled DataFrame
df1.rename(columns={'RollNo':'RegNo'},inplace=True)
print(df1)
```

```python
# Renaming multiple columns in a labelled DataFrame
df1.rename(columns={'RegNo':'RollNo','Name':'StuName'},inplace=True)
print(df1)
```

```python
# Renaming Single Row in a labelled DataFrame
df1.rename(index={'A1':'A21'},inplace=True)
print(df1)
```

```python
# Renaming multiple row in a labelled DataFrame
df1.rename(index={'A2':'A22','A3':'A23'},inplace=True)
print(df1)
```

```python
print(df2)
```

```python
# Renaming system-assigned column names to user-defined column names
# Process is similar for single/multiple columns
df2.rename(columns={0:'RegNo',1:'CustName'},inplace=True)
print(df2)
```

```python
# Renaming system-assigned index values to user-defined index values.
# Process is similar for single/multiple indices.
df2.rename(index={0:'C1',1:'C2'},inplace=True)
print(df2)
```

# Iterations in a DataFrame

```python
df1
```

```python
for element in df1:
    print(element)
```

```python
# Iteration over rows
for row_index, row_value in df1.iterrows():
    print(row_index)
    print('-'*10)
    print(row_value)
```

```python
# Iteration over columns
for col_index, col_value in df1.items():
    print(col_index)
    print('-'*10)
    print(col_value)
```
