```python
import pandas as pd
import numpy as np
```

```python
data = {
    'ProductID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductName': ['iPhone 15', 'Galaxy S23', 'MacBook Air', 'Dell XPS', 'AirPods',
                    'Sony WH-1000XM5', 'iPad Pro', 'Kindle', 'Apple Watch', 'Gaming Mouse'],
    'Category': ['Mobile', 'Mobile', 'Laptop', 'Laptop', 'Audio', 'Audio', 'Tablet', 'E-Reader', 'Wearable', 'Accessories'],
    'Brand': ['Apple', 'Samsung', 'Apple', 'Dell', 'Apple', 'Sony', 'Apple', 'Amazon', 'Apple', 'Logitech'],
    'Price': [799, 749, 999, 1200, 249, 349, 899, 139, 399, 79],
    'Quantity': [15, 10, 8, 5, 50, 12, 0, 20, 18, 25],
    'Rating': [4.8, 4.5, 4.9, 4.2, 4.7, 4.8, 4.6, 4.4, 4.5, 4.1]
}

inventory_df = pd.DataFrame(data)
```

```python
df1 = inventory_df.copy()
```

```python
apple_condition = (df1['Brand'] == 'Apple') & (df1['Category'].isin(['Mobile', 'Wearable']))

apple_products = df1[apple_condition]
print(apple_products)
```

```python
import numpy as np

# 1. Define the list of conditions
conditions = [
    (df1['Quantity'] > 20),                                # Condition 1
    (df1['Quantity'] >= 1) & (df1['Quantity'] <= 20), # Condition 2
    (df1['Quantity'] == 0)                                 # Condition 3
]

# 2. Define the corresponding choices for each condition
choices = ['High Stock', 'Available', 'Out of Stock']

# 3. Use np.select to assign the values based on conditions
df1['Status'] = np.select(conditions, choices, default='Unknown')

# Display the updated DataFrame
print(df1[['ProductName', 'Quantity', 'Status']])
```
