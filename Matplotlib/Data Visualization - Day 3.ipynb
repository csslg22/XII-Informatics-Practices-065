## Histogram
### Introduction.

A histogram is a graphical representation which organizes a group of data points into user-specified ranges. Histogram provides a visual interpretation of numerical data by showing the number of data points that fall within a specified range of values ("bins"). It is similar to a vertical bar graph but without gaps between the bars.

### Simple Histogram

```python
import matplotlib.pyplot as plt

marks = [12, 18, 22, 24, 27, 31, 35, 39, 42, 
         48, 53, 56, 61, 65, 72, 78]

plt.hist(marks)

plt.show()
```

**Explanation:**

* `marks` contains the observations.

* Since `bins` are not specified, Matplotlib automatically creates suitable intervals.

* The height of each bar represents the number of observations in that interval.

### Understanding the Bins (Intervals)

A histogram groups data into **intervals** called **bins**.

Suppose the bins are:

```python
bins=[0,10,20,30,40,50,60]
```

The correct version should be:


| Bin Number | Interval |
| ---------- | -------- |
| 1          | 0–10     |
| 2          | 10–20    |
| 3          | 20–30    |
| 4          | 30–40    |
| 5          | 40–50    |
| 6          | 50–60    |


Every value is placed into one of these intervals.

### Specifying Bins

```python
import matplotlib.pyplot as plt

marks = [12,18,22,24,27,31,35,39,42,48,53,56]

plt.hist(marks,
         bins=[0,10,20,30,40,50,60])

plt.show()
```

**Explanation**

Values are grouped as follows:

| Interval | Values   |
| -------- | -------- |
| 0–10     | None     |
| 10–20    | 12,18    |
| 20–30    | 22,24,27 |
| 30–40    | 31,35,39 |
| 40–50    | 42,48    |
| 50–60    | 53,56    |


The bar height equals the frequency (number of values).

### Understanding the `weights` parameter

Normally, the height of each bar is the **frequency**.

The `weights` parameter allows us to assign custom heights to bars.

Instead of counting observations, Matplotlib **adds the weights** of observations that fall into the same bin.

This is extremely useful when each observation represents a different quantity.

**Histogram with Weights**

```python
import matplotlib.pyplot as plt

positions = [5,15,25,35,45,55]

weights = [20,10,45,33,6,8]

plt.hist(positions,
         bins=[0,10,20,30,40,50,60],
         weights=weights)

plt.show()
```

**How it works**

Each position falls into a different interval.

| Position | Bin   | Weight |
| -------- | ----- | ------ |
| 5        | 0–10  | 20     |
| 15       | 10–20 | 10     |
| 25       | 20–30 | 45     |
| 35       | 30–40 | 33     |
| 45       | 40–50 | 6      |
| 55       | 50–60 | 8      |


Therefore the histogram displays:

| Interval | Bar Height |
| -------- | ---------- |
| 0–10     | 20         |
| 10–20    | 10         |
| 20–30    | 45         |
| 30–40    | 33         |
| 40–50    | 6          |
| 50–60    | 8          |


**Important**

The number of values in the first list and the `weights` list must be equal, otherwise Python generates an error.

### Multiple Values in the Same Bin

```python
import matplotlib.pyplot as plt

positions = [5,15,25,35,15,55]

weights = [20,10,45,33,6,8]

plt.hist(positions,
         bins=[0,10,20,30,40,50,60],
         weights=weights)

plt.show()
```

**Explanation**

Notice that **15 appears twice**.6

The corresponding weights are: 10 6

Both values belong to the interval: 10-20

Therefore, `Bar Height = 10 + 6 = 16`

The histogram becomes:

| Interval | Height |
| -------- | ------ |
| 0–10     | 20     |
| 10–20    | 16     |
| 20–30    | 45     |
| 30–40    | 33     |
| 40–50    | 0      |
| 50–60    | 8      |


This example clearly demonstrates that weights are added within the same interval.

## Customizing Histogram

**Changing Bar Color**

```python
import matplotlib.pyplot as plt

marks = [12,18,22,24,27,31,35,39,42,48,53]

plt.hist(marks,
         bins=[0,10,20,30,40,50,60],
         color="orange")

plt.show()
```

**Changing Edge Color**

```python
plt.hist(marks,
         bins=[0,10,20,30,40,50,60],
         color="skyblue",
         edgecolor="red")
plt.show()
```

**Changing Edge Width**

```python
plt.hist(marks,
         bins=[0,10,20,30,40,50,60],
         edgecolor="black",
         linewidth=3)
plt.show()
```

**Make Bars Transparent**

```python
plt.hist(marks,
         bins=[0,10,20,30,40,50,60],
         color="green",
         alpha=0.2)
plt.show()
```

`alpha` ranges from: 

0 → Completely transparent

1 → Completely opaque

**Add Title, X-Label, Y-Label**

```python
plt.hist(marks,
         bins=[0,10,20,30,40,50,60],
         color="skyblue",
         edgecolor="red")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Student Marks")
plt.show()
```

**Changing Figure Size**

```python
plt.figure(figsize=(8,4))
plt.hist(marks)
plt.show()
```

**Changing Bin Width**

Small Bins

```python
plt.hist(marks,bins=12)
plt.show()
```

Large Bins

```python
plt.hist(marks,bins=5)
plt.show()
```

Smaller bins **give more detail**.

Larger bins **give a simpler overview**.

**Horizontal Histogram**

```python
plt.hist(marks,
         bins=[0,10,20,30,40,50,60],
         orientation="horizontal")
plt.show()
```

**Use Different Histogram Styles**

```python
plt.hist(marks,
         bins=6,
         histtype="step")
plt.show()
```

Other styles:
* bar
* barstacked
* step
* stepfilled

```python
plt.hist(marks,
         bins=6,
         histtype="stepfilled")
plt.show()
```

**Display Density Instead of Frequency**

```python
plt.hist(marks,
         bins=6,
         density=True)
plt.show()
```

The total area under the histogram becomes 1, which is useful for probability distributions.

**Align the Bars**

```python
plt.hist(marks,
         bins=6,
         align="mid")
plt.show()
```

Options: `left`, `mid`, `right`

**Cumulative Histogram**

```python
plt.hist(marks,
         bins=6,
         cumulative=True)
plt.show()
```

Each bar shows the cumulative frequency.

**Add Grid Lines**

```python
plt.hist(marks,
         bins=6,
         cumulative=True)
plt.grid(True)
plt.show()
```

**Add Multiple Histograms**

```python
import matplotlib.pyplot as plt

boys = [42,50,52,60,65,70,75,80]
girls = [38,45,49,55,60,66,72,78]

plt.hist([boys,girls],
         bins=[30,40,50,60,70,80,90],
         label=["Boys","Girls"])

plt.legend()

plt.show()
```

**Complete Example**

```python
import matplotlib.pyplot as plt

marks = [12,18,22,24,27,31,35,39,42,48,53,56]

plt.figure(figsize=(8,5))

plt.hist(
    marks,
    bins=[0,10,20,30,40,50,60],
    color="skyblue",
    edgecolor="black",
    linewidth=2,
    alpha=0.8
)

plt.title("Distribution of Student Marks",fontsize=15)

plt.xlabel("Marks",fontsize=12)

plt.ylabel("Frequency",fontsize=12)

plt.grid(axis="y",linestyle="--",alpha=0.6)

plt.show()
```

**Frequently Used Parameters of `plt.hist()`**

| Parameter     | Purpose                                         |
| ------------- | ----------------------------------------------- |
| `x`           | Data values                                     |
| `bins`        | Defines the intervals (bins)                    |
| `weights`     | Assigns custom weights to observations          |
| `color`       | Fill color of bars                              |
| `edgecolor`   | Color of bar borders                            |
| `linewidth`   | Thickness of borders                            |
| `alpha`       | Transparency of bars                            |
| `histtype`    | Histogram style (`bar`, `step`, etc.)           |
| `orientation` | Vertical or horizontal histogram                |
| `density`     | Normalizes histogram so the total area equals 1 |
| `cumulative`  | Shows cumulative frequencies                    |
| `align`       | Alignment of bars within bins                   |
| `label`       | Label used in the legend                        |
| `rwidth`      | Relative width of bars (0 to 1)                 |


## Saving the Figure

```python
import matplotlib.pyplot as plt

boys = [42,50,52,60,65,70,75,80]
girls = [38,45,49,55,60,66,72,78]

plt.hist([boys,girls],
         bins=[30,40,50,60,70,80,90],
         label=["Boys","Girls"])
plt.legend()
plt.savefig('Histogram.png') # Should be placed before plt.show()
plt.show()
```

For future use we have to save the plot.To save any plot `savefig()`
method is used. Plots can be saved like `pdf`, `svg`, `png`, `jpg` file
formats.ts

The figure gets saved to the current working directory. The method is applicable for all the types of charts.
