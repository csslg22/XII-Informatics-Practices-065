## Bar Graph

### Introduction

A graph drawn using rectangular bars to show how
large each value is. The bars can be horizontal or
vertica

A bar graph makes it easy to compare data
between different groups at a glance. Bar graph
represents categories on one axis and a discrete
value in the other. The goal bar graph is to show
the relationship between the two axes. Bar graph
can also show big changes in data over time.

A Simple Bar Graph

````python
import matplotlib.pyplot as plt
import numpy as np
label = ['Anil', 'Vikas', 'Dharma', 'Mahen', 'Manish', 'Rajesh']
per = [94,85,45,25,50,54]
index = np.arange(len(label))
plt.bar(index, per)
plt.xlabel('Student Name')
plt.ylabel('Percentage')
plt.xticks(index, label)
plt.title('Percentage of Marks achieve by student Class XII')
plt.show()
````

Horizontal Bar Graph

````python
import matplotlib.pyplot as plt
import numpy as np

# Student names
label = ['Anil', 'Vikas', 'Dharma', 'Mahen', 'Manish', 'Rajesh']

# Percentage marks
per = [94, 85, 45, 25, 50, 54]

# Position of bars
index = np.arange(len(label))

# Horizontal bar graph
plt.barh(index, per,
         color="orange",
         edgecolor="blue",
         linewidth=2,
         linestyle="--",
         label="Class XII")

# Labels
plt.xlabel("Percentage (%)", fontsize=10)
plt.ylabel("Student Names", fontsize=10)

# Display student names on Y-axis
plt.yticks(index, label, fontsize=9)

# Title
plt.title("Class XII Students' Percentage (Horizontal Bar Graph)")

# Legend
plt.legend(loc="lower right", frameon=True)

# Display graph
plt.show()
````

Multiple Bar Plots

A multiple bar plot (grouped bar chart) is used to compare two or more datasets for the same categories. In this example, we compare the Half-Yearly and Annual exam percentages of students.

````python
import matplotlib.pyplot as plt
import numpy as np

# Student names
students = ['Anil', 'Vikas', 'Dharma', 'Mahen', 'Manish', 'Rajesh']

# Marks in two exams
half_yearly = [85, 78, 60, 45, 70, 68]
annual = [94, 85, 45, 25, 50, 54]

# Position of bars
index = np.arange(len(students))
width = 0.35

# First bar graph
plt.bar(index, half_yearly,
        width=width,
        color='skyblue',
        edgecolor='black',
        label='Half-Yearly')

# Second bar graph
plt.bar(index + width, annual,
        width=width,
        color='orange',
        edgecolor='black',
        label='Annual')

# Labels and title
plt.xlabel('Student Name', fontsize=10)
plt.ylabel('Percentage', fontsize=10)
plt.title('Comparison of Half-Yearly and Annual Exam Marks')

# Student names on X-axis
plt.xticks(index + width/2, students, rotation=30)

# Legend
plt.legend(loc='upper right', frameon=True)

# Display graph
plt.show()
````

**Explanation of the Code**

* `index = np.arange(len(students))` creates an array of positions for the first set of bars on the x-axis, with one position for each student.

* `width = 0.35` defines the width of each bar. This value also determines the spacing between the two bars representing each student.

* `plt.bar(index, half_yearly, width, label='Half-Yearly')` plots the first set of bars representing the **Half-Yearly** exam marks.

* `plt.bar(index + width, annual, width, label='Annual')` plots the second set of bars representing the **Annual** exam marks. Adding `width` to `index` shifts these bars slightly to the right so that both bars are displayed side by side.

* `plt.xticks(index + width/2, students)` places the student names at the center of each pair of bars, making the chart easier to read.

* `plt.legend()` displays a legend that identifies the two datasets: **Half-Yearly** and ***ual**.**

### Result

The output is a **grouped (multiple) bar chart** in which each student has two adjacent bars—one representing the **Half-Yearly** marks and the other representing the **Annual** marks. This visualization makes it easy to compare the performance of students across the two examinations.


````python
import matplotlib.pyplot as plt
import numpy as np

# Student names
students = ['Anil', 'Vikas', 'Dharma', 'Mahen', 'Manish', 'Rajesh']

# Marks in three examinations
unit_test = [78, 72, 55, 40, 62, 58]
half_yearly = [85, 78, 60, 45, 70, 68]
annual = [94, 85, 45, 25, 50, 54]

# Position of bars
index = np.arange(len(students))
width = 0.25

# Plot three bar graphs
plt.bar(index - width, unit_test,
        width=width,
        color='skyblue',
        edgecolor='black',
        label='Unit Test')

plt.bar(index, half_yearly,
        width=width,
        color='lightgreen',
        edgecolor='black',
        label='Half-Yearly')

plt.bar(index + width, annual,
        width=width,
        color='orange',
        edgecolor='black',
        label='Annual')

# Labels and title
plt.xlabel('Student Name', fontsize=10)
plt.ylabel('Percentage (%)', fontsize=10)
plt.title('Comparison of Student Performance in Three Examinations')

# Display student names on X-axis
plt.xticks(index, students, rotation=30)

# Legend
plt.legend(loc='upper right', frameon=True)

# Display graph
plt.show()
````

**Explanation**

- `width = 0.25` sets the width of each bar.
- `index - width` draws the **Unit Test** bars on the left.
- `index` draws the **Half-Yearly** bars in the center.
- `index + width` draws the **Annual** bars on the right.
- `plt.xticks(index, students)` places the student names below the center of each group.
- `plt.legend()` displays the names of the three datasets.

The grouped bar chart displays three bars for each student, making it easy to compare their performance across the Unit Test, Half-Yearly, and Annual examinations.

````python
import matplotlib.pyplot as plt
import numpy as np

# Student names
students = ['Anil', 'Vikas', 'Dharma', 'Mahen', 'Manish', 'Rajesh']

# Marks in three examinations
unit_test = [78, 72, 55, 40, 62, 58]
half_yearly = [85, 78, 60, 45, 70, 68]
annual = [94, 85, 45, 25, 50, 54]

# Position of bars
index = np.arange(len(students))
height = 0.25

# Plot three horizontal bar graphs
plt.barh(index - height, unit_test,
         height=height,
         color='skyblue',
         edgecolor='black',
         label='Unit Test')

plt.barh(index, half_yearly,
         height=height,
         color='lightgreen',
         edgecolor='black',
         label='Half-Yearly')

plt.barh(index + height, annual,
         height=height,
         color='orange',
         edgecolor='black',
         label='Annual')

# Labels and title
plt.xlabel('Percentage (%)', fontsize=10)
plt.ylabel('Student Name', fontsize=10)
plt.title('Comparison of Student Performance in Three Examinations')

# Display student names on Y-axis
plt.yticks(index, students)

# Legend
plt.legend(loc='upper right', frameon=True)

# Display graph
plt.show()
````

The `barh()` function in **Matplotlib** is used to draw **horizontal bar charts**. It is similar to `bar()`, but the bars are displayed horizontally instead of vertically.

The syntax of `barh()` is:

```python
plt.barh(y, width, height=0.8, color=None,
         edgecolor=None, label=None)
```

**Parameters:**

- **y** – Specifies the position of the bars on the y-axis.
- **width** – Represents the length of each horizontal bar (data values).
- **height** – Specifies the thickness of each bar.
- **color** – Sets the fill color of the bars.
- **edgecolor** – Sets the border color of the bars.
- **label** – Provides a name for the legend.

The program works as follows:

1. Import the required libraries:
   - `matplotlib.pyplot` for plotting graphs.
   - `numpy` for creating bar positions.

2. Store the student names and their marks in three examinations:
   - Unit Test
   - Half-Yearly
   - Annual

3. Create bar positions using:

   ```python
   index = np.arange(len(students))
   ```

   This generates equally spaced positions for each student.

4. Set the thickness of each horizontal bar:

   ```python
   height = 0.25
   ```

5. Draw three horizontal bar charts using `plt.barh()`:
   - `index - height` → Unit Test
   - `index` → Half-Yearly
   - `index + height` → Annual

   These offsets ensure that the three bars for each student are displayed side by side without overlapping.

6. Add labels to the axes and give a title to the chart:

   ```python
   plt.xlabel('Percentage (%)')
   plt.ylabel('Student Name')
   plt.title('Comparison of Student Performance in Three Examinations')
   ```

7. Display the student names on the y-axis:

   ```python
   plt.yticks(index, students)
   ```

8. Display the legend:

   ```python
   plt.legend()
   ```

9. Display the graph:

   ```python
   plt.show()
   ```

The main differences between `bar()` and `barh()` are:

| `bar()` (Vertical Bar Chart) | `barh()` (Horizontal Bar Chart) |
|------------------------------|----------------------------------|
| Bars are drawn vertically. | Bars are drawn horizontally. |
| Categories appear on the X-axis. | Categories appear on the Y-axis. |
| Values are shown on the Y-axis. | Values are shown on the X-axis. |
| Uses `width` to set bar thickness. | Uses `height` to set bar thickness. |
| Uses `plt.xticks()` for category labels. | Uses `plt.yticks()` for category labels. |
