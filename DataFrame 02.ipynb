# Boolean Indexing and Filtering in DataFrame

```python
import numpy as np
data = {
    'Name'  : ['Amit', 'Bhavna', 'Deepa', 'Esha'],
    'Marks' : [45, 85, 92, 55]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
print(df['Marks']>80) # Gives us a Boolean Series output
```

```python
print(df[df['Marks']>80]) # Gives us the actual values which satisfy the condition
```

```python
# Create a column 'Grade':
'''
Marks >= 90 -> 'A'
Marks >= 70 but <= 89 -> 'B'
Marks >= 50 but <= 69 -> 'C'
Marks < 50 -> 'D'
'''

conditions = [
    (df['Marks'] >= 90),
    ((df['Marks'] >= 70) & (df['Marks'] <= 89)),
    ((df['Marks'] >= 50) & (df['Marks'] <= 69)),
    (df['Marks'] < 50)
]

choices = ['A', 'B', 'C', 'D']

df['Grade'] = np.select(conditions, choices, default='D')
df
```
