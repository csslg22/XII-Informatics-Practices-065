# Boolean Indexing and Filtering in DataFrame

```python
import numpy as np
data = {
    'Name'  : ['Amit', 'Bhavna', 'Deepa', 'Esha'],
    'Marks' : [45, 85, 92, 55]
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
print(df['Marks']>80) # Gives us a Boolean Series output
```

```python
print(df[df['Marks']>80]) # Gives us the actual values which satisfy the condition
```

```python
import pandas as pd
import numpy as np

data = {
    'Name'     : ['Amit','Bhavna','Chintu','Deepa','Esha',
                   'Faisal','Gaurav','Hina','Ishan','Sara','Ravi'],
    'Class'    : [11, 12, 11, 11, 12, 11, 12, 11, 12, 12, 11],
    'Section'  : ['A','B','A','C','A','B','A','B','C','A','C'],
    'Marks'    : [45, 85, 30, 92, 55, 62, 78, 25, 48, 95, 60],
    'Physics'  : [50, 80, 20, 95, 60, 65, 80, 30, 40, 90, 55],
    'Chemistry': [40, 85, 35, 90, 50, 70, 75, 20, 55, 92, 65],
    'English'  : [60, 75, 40, 88, 65, 55, 85, 45, 60, 89, 70],
    'Remarks'  : ['Average','Good','Poor','Excellent','Average',
                   'Average','Good','Poor','Average','Excellent','Average']
}
df = pd.DataFrame(data)
```

```python
print(df)
```

```python
df
```

```python
print(df[df['Marks']>80])
```

```python
print(df[df['Physics']<60])
```

```python
print(df[(df['Chemistry']>=50) & (df['Chemistry']<=70)])
```

```python
print(df[df['Remarks']=='Poor'])
```

```python
print(df[(df['Class']==12) & (df['Section']=='C')])
```

```python
print(df[(df['Class']==11) | (df['Section']=='C')])
```

```python
print(df[~(df['Class'] == 11)])
```

```python
print(df.shape)
```

```python
# Display the Name, Class and Section of those students whose marks >= 75
print(df.loc[(df['Marks']>=75),['Name','Class','Section']])
```

```python
print(df.loc[(df['Marks']>=75),'Name'])
```

```python
# Display the name and marks of Physics of those students whose Remarks is Average

print(df.loc[(df['Remarks']=='Average'),['Name', 'Physics']])
```

```python
print(df.loc[(df['Remarks']=='Average'),:])
```

```python
# Display the name, class and section of those students whose marks in Physics is greater than 70 and Marks in English is greater than 50.

print(df.loc[(df['Physics']>70) & (df['English']>50),['Name','Class','Section']])
```

```python
# Display the name and remarks of those students whose section is not B

print(df.loc[(df['Section']!='B'),['Name','Section','Remarks']])
```

```python
print(df.loc[~(df['Section']=='B'),['Name','Section','Remarks']])
```
